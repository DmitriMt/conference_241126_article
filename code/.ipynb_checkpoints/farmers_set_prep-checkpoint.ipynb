```python
def extract_from_dict_key(dict_, search_word):
    for key, val in dict_.items():
        if search_word.lower() in val.lower() or search_word.lower() in key.lower():
            print(f"{key}: {val}")
```

```python
farmers_df = df_pl.with_columns(
    (
        (pl.col("j90") == 10)
        | ((pl.col("j4_1") == 8) & (pl.col("j11").is_in([2, 3, 4])))
        | (
            (pl.col("j56") == 1)
            & ((pl.col("j60_4a2") == 1) | (pl.col("j60_4a3") == 1))
            & (pl.col("j57") > 10_000)
        )
    ).alias("is_farmer")
).with_columns(
    pl.when(pl.col("is_farmer") == True).then(1).otherwise(0).alias("is_farmer")
)
farmers_df = farmers_df.filter(pl.col("year") >= 2015)
farmers_df
```

```python
df_2025_complete = farmers_df.to_pandas()
df_2025_complete["gender"] = df_2025_complete["h5"].replace(
    meta.variable_value_labels["h5"]
)
df_2025_complete["is_female"] = np.where(df_2025_complete["gender"] == "FEMALE", 1, 0)
df_2025_complete["diplom"] = df_2025_complete["diplom"].replace(
    meta.variable_value_labels["diplom"]
)
df_2025_complete["marst"] = df_2025_complete["marst"].replace(
    meta.variable_value_labels["marst"]
)
df_2025_complete["region"] = df_2025_complete["region"].replace(
    meta.variable_value_labels["region"]
)
df_2025_complete["i3"] = df_2025_complete["i3"].replace(
    meta.variable_value_labels["i3"]
)
df_2025_complete["j90"] = df_2025_complete["j90"].replace(
    meta.variable_value_labels["j90"]
)
df_2025_complete["j4_1"] = df_2025_complete["j4_1"].replace(
    meta.variable_value_labels["j4_1"]
)
df_2025_complete["i4"] = df_2025_complete["i4"].replace(
    meta.variable_value_labels["i4"]
)
df_2025_complete["j1_1_1"] = df_2025_complete["j1_1_1"].replace(
    meta.variable_value_labels["j1_1_1"]
)
df_2025_complete["j1_1_5"] = df_2025_complete["j1_1_5"].replace(
    meta.variable_value_labels["j1_1_5"]
)
df_2025_complete["j1_1_3"] = df_2025_complete["j1_1_3"].replace(
    meta.variable_value_labels["j1_1_3"]
)
df_2025_complete["j1_1_6"] = df_2025_complete["j1_1_6"].replace(
    meta.variable_value_labels["j1_1_6"]
)
df_2025_complete["j60_4a2"] = df_2025_complete["j60_4a2"].replace(
    meta.variable_value_labels["j60_4a2"]
)
df_2025_complete["m20_7"] = df_2025_complete["m20_7"].replace(
    meta.variable_value_labels["m20_7"]
)
df_2025_complete = df_2025_complete.rename(
    columns={
        "j60": "total_income_last30_days",
        'j10': 'job1_wage',
        'j40':'job2_wage',
        'j57':'iea_income',
        'j363': 'pension_benefit', 
        "j72_172": "children_num",
        "i3": "settlement_type_born",
        "h3": "family_id",
        "h4": "family_member_id",
        "j90": "primary_occupation",
        "j4_1": "industry_job1",
        'j35_1': 'industry_job2',
        "j60_4a2": "raised_and_sell_crops",
        "j8": "ow",
        "j38": "ow2",
        "j59": "ow_iea",
        "j1_1_1": "job_satisfaction",
        'j1_1_3': 'satisfaction_w_pay',
        'j1_1_6': 'satisfaction_w_ow',
        'j1_1_5': 'satisfaction_w_life',
        'i4': 'nationality',
        'm20_7': 'disability_classs'
        
    }
)
df_2025_complete = df_2025_complete[
    [
        'year',
        "family_id",
        "family_member_id",
        "total_income_last30_days",
        'job1_wage',
        'job2_wage',
        'iea_income',
        'pension_benefit',
        "age",
        "gender",
        "is_female",
        "diplom",
        "marst",
        "children_num",
        "settlement_type_born",
        "region",
        "primary_occupation",
        "industry_job1",
        'industry_job2',
        "raised_and_sell_crops",
        "ow",
        "ow2",
        "ow_iea",
        'job_satisfaction',
        'satisfaction_w_pay',
        'satisfaction_w_ow',
        'satisfaction_w_life',
        "is_farmer",
        'nationality',
        'disability_classs',
        "inwgt",
    ]
]
df_2025_complete["total_income_last30_days"] = np.where(
    df_2025_complete["total_income_last30_days"].isin(
        [99999999.0, 99999998.0, 99999997.0]
    ),
    0,
    df_2025_complete["total_income_last30_days"],
)
df_2025_complete["ow"] = np.where(
    df_2025_complete["ow"].isin([99999999.0, 99999998.0, 99999997.0]),
    np.nan,
    df_2025_complete["ow"],
).astype(float)
df_2025_complete["ow2"] = np.where(
    df_2025_complete["ow2"].isin([99999999.0, 99999998.0, 99999997.0]),
    np.nan,
    df_2025_complete["ow2"],
).astype(float)
df_2025_complete["ow_iea"] = np.where(
    df_2025_complete["ow_iea"].isin([99999999.0, 99999998.0, 99999997.0]),
    np.nan,
    df_2025_complete["ow_iea"],
).astype(float)
df_2025_complete["hours_worked"] = (
    df_2025_complete[["ow", "ow2", "ow_iea"]].bfill(axis=1).iloc[:, 0]
)
df_2025_complete["disability_classs"] = np.where(
    df_2025_complete["disability_classs"].isin(['Yes', 'DISABLED CHILD [FOR RESPONDENTS UNDER 18 YEARS OLD]', 'DOING THE PAPERWORK']),
    1,
    0,
).astype(float)

df_2025_complete
```

```python
def reconstruct_families(
    df, member_col="family_member_num", new_family_col="new_family_id"
):
    """
    Reconstruct family groupings based on member enumeration resets.

    Parameters:
    -----------
    df : pandas.DataFrame
        DataFrame containing the enumerated family members
    member_col : str
        Column name with the enumerated family member numbers
    new_family_col : str
        Name for the new reconstructed family ID column

    Returns:
    --------
    pandas.DataFrame with new family_id and family_size columns
    """
    df = df.copy()

    # Detect where a new family starts:
    # A reset occurs when current value <= previous value
    # First row always starts a new family
    is_new_family = (df[member_col].diff() <= 0) | (df[member_col].diff().isna())

    # Cumulative sum creates a new group id for each reset
    df[new_family_col] = is_new_family.cumsum()

    # Compute total members per family
    family_sizes = df.groupby(new_family_col)[member_col].transform("size")
    df["family_size"] = family_sizes

    return df
```

```python
df_2025_complete = reconstruct_families(
    df_2025_complete, member_col="family_member_id", new_family_col="family_id_imputed"
)
df_2025_complete
```
